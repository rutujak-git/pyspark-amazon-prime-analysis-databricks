### B] Data Cleaning

#### 1. Handle missing values

```
df = df.na.fill({'rating': 'Unrated', 'country': 'Unknown'})
df.display()
```

```
df = df.dropna('any')
df.display()
```

```
df = df.dropDuplicates()
df.display()
```

```
df = df.withColumnRenamed('title', 'content_title')
df.display()
```

### C] Data Transformation

#### Add new columns and transform existing data for deeper insights

```
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```
df = df.withColumn('is_movie', when(col('type') == 'Movie', 1).otherwise(0))  ## Creates a new column is_movie with a value of 1 for rows where the type is 'Movie', and 0 otherwise.
df.display()
```

```
df = df.withColumn('is_series', when(col('type') == 'TV Show', 1).otherwise(0)) 
df.display()
```

#### Analyze genres by splitting and counting occurrences

```
# Analyze categories by splitting and counting occurrences in the 'listed_in' column
df = df.withColumn('categories', split(col('listed_in'), ', '))  # Split 'listed_in' into an array of categories
category_analysis = (
    df.select(explode(col('categories')).alias('category'))  # Explode the categories array into individual rows
    .groupBy('category')  # Group by category
    .count()  # Count the number of occurrences for each category
    .orderBy('count', ascending=False)  # Order by count in descending order
)

# Display category analysis
category_analysis.display()
```

#### Part 4: Exploratory Data Analysis (EDA)

```
df.groupBy('type').count().orderBy('count', ascending=False).display()  # Command: Groups data by 'type' (Movie/TV Show) and counts occurrences
```

```
df.groupBy('release_year').count().orderBy('release_year').display()  # Command: Groups data by release year and counts occurrences
```

```
df.select(explode(col('categories')).alias('category')).groupBy('category').count().orderBy('count', ascending=False).display()  # Command: Explodes 'categories' and counts occurrences
```

### Part 5: Business Insights

```
df.filter(df['rating'] != 'Unrated').orderBy(col('rating').desc()).display()  # Command: Filters and shows top-rated content
```

```
df.groupBy('country').count().orderBy('count', ascending=False).display()  # Command: Groups by country and counts occurrences
```

```
df.groupBy('director').count().orderBy('count', ascending=False).display()  # Command: Groups by director and counts occurrences
```

### Part 6: Advanced Analytics

```
df.sort(col('release_year').desc(), col('rating').desc()).display()  # Command: Sorts by release year and rating in descending order
```

```
from pyspark.sql.window import Window
from pyspark.sql.functions import row_number
```

```
# Define window specification
window_spec = Window.partitionBy('release_year').orderBy(col('rating').desc())
```

```
# Apply window function to rank content within each year
df = df.withColumn('rank', row_number().over(window_spec))
```

```
# Show top 5 content for each year
df.filter(df['rank'] <= 5).display()  # Command: Ranks content within each year and shows top 5
```

### Part 7: Data Export

```
# Write the cleaned and transformed data to Delta format
df.write.format('delta').mode('overwrite').save('/FileStore/processed_data/')  # Command: Writes the data to Delta format
```

```
# Export the data as CSV
df.write.format('csv').mode('overwrite').save('/FileStore/processed_data_csv/')  # Command: Writes the data to CSV format
```

```
from pyspark.sql.functions import concat_ws

# Convert the 'categories' array column into a string of comma-separated values
df = df.withColumn('categories', concat_ws(', ', col('categories')))  # Command: Converts array to string

# Write the data to CSV
df.write.format('csv').mode('overwrite').save('/FileStore/processed_data_csv/')  # Command: Writes the data to CSV format
```
